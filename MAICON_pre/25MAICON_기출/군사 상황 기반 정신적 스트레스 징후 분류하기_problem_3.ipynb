# 병사 발언 분석을 통한 스트레스 징후 탐지하기

본 임무의 목적은 주어진 병사의 발언을 분석하여 스트레스 징후를 예측하는 분류기를 만드는 것입니다. 본 과제에서 ‘병사’는 현역과 전역자를 모두 포괄합니다. 스트레스 징후를 파악하는 것이 중요하기에 평가는 단순 전체 정확도가 아닌 스트레스 징후를 나타내는 데이터를 기준으로 F1 점수를 계산하여 진행합니다.

## 훈련 데이터

`train.csv`는 두 개의 필드로 구성되며 총 300개의 레코드가 제공됩니다.

| 필드 | 타입 | 설명 |
| --- | --- | --- |
| text | 문자열 | 병사 발언에 해당하는 짧은 한국어 문장이 들어옵니다. 오타가 존재할 수 있고 띄어쓰기가 안 맞을 수도 있습니다. |
| stress | 숫자 | 예측 대상이 되는 레이블 필드로 1이면 스트레스 징후가 있는 것이고 0이면 정상입니다 |

처음 필드 text는 병사의 말이고 stress는 그 text에 스트레스 가능성을 표현하는지 여부로 1이면 스트레스 가능성이 있다고 간주합니다. 0이면 정상입니다. 아래 처음 5개의 레코드를 예로 보여드렸습니다.

| text | stress |
| --- | --- |
| 금속 부딪히는 소리에 가슴이 철렁해 한동안 걸음을 멈췄다.| 1 |
| 부두 경계 근무 때로는 긴장되지만 밤에 깨더라도 금세 다시 잔다 | 0 |
| 아팠던 몸도 충분한 휴식 후에는 완전히 회복됐다 | 0 |
| 장갑차 운전 교육을 받을 때 큰 차체와 복잡한 조작법 때문에 두려웠지만 교관의 세심한 지도와 충분한 연습을 통해 안전하고 정확한 기동이 가능해졌다 | 0 |
| 그뒤로 훈련 중 폭죽 소리 때로는 긴장되지 만경계는 하되 업무는 평소대로 한다 | 0 |

이 데이터셋을 이용해 직접 분류기 모델을 학습(train) 하고 평가해야 합니다. 단, GPT와 같은 대규모 언어 모델(LLM)을 직접 불러와 Zero-Shot 또는 Few-Shot과 같은 프롬프트 엔지니어링만으로 분류 작업을 수행하는 것은 허용되지 않습니다. 즉, LLM을 단순 분류기로 호출하는 것은 금지하며, **반드시 제공된 데이터셋을 기반으로 별도의 모델을 학습시켜야 합니다.** 모델을 명확히 학습해야 하기에 매뉴얼하게 직접 분류하여 제출하는 것 역시 허용하지 않습니다. 필요시 작성 코드를 확인할 예정입니다.

## 평가 기준

앞서 언급했던 것처럼 스트레스 징후를 찾아내는 것이 더 중요하기에 스트레스를 나타내는 레이블 1(스트레스 징후 있음)을 기준으로 f1 스코어를 계산합니다. 레이블 1을 기준으로 정밀도(precision)와 재현율(recall)을 계산한 다음에 아래와 같은 공식으로 f1 스코어를 계산합니다.

* precision = 모델이 "스트레스 징후 있음" 이라고 예측한 것 중에서 실제로 맞은 비율
* recall = 실제로 "스트레스 징후 있음" 인 것 중에서 모델이 제대로 찾아낸 비율
* f1 = 2 * precision * recall / (precision + recall)

```python
# MAICON 2025 예선 문제 3 - 병사 발언 스트레스 징후 분류 (F1, label=1 기준)
# 모델: [char 1~3gram TF-IDF + 감정 어휘 피처 → LR] + [char_wb 1~4gram TF-IDF → LR] 확률 평균 앙상블
import re, warnings
import numpy as np, pandas as pd
from scipy import sparse
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline, make_union
from sklearn.preprocessing import FunctionTransformer
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.metrics import f1_score, precision_score, recall_score
warnings.filterwarnings("ignore")

TRAIN_PATH, TEST_PATH, OUT_PATH = "./problem_3_data/data/train.csv", "./problem_3_data/data/test.csv", "./problem_3_data/submission.csv"
SEED, C, THRESHOLD = 42, 10, 0.46

# ---------- 1. 데이터 ----------
def clean(s):  # 연속 공백/말줄임 기호 정리 (띄어쓰기 오류 완화)
    s = str(s).replace("…", " ").replace("..", " ")
    return re.sub(r"\s+", " ", s).strip()

train = pd.read_csv(TRAIN_PATH); test = pd.read_csv(TEST_PATH)
X, y, X_test = train.text.map(clean).values, train.stress.astype(int).values, test.text.map(clean).values

# ---------- 2. 감정 어휘 피처 (학습 모델의 입력 피처로만 사용) ----------
NEG = ['불안','두렵','두려','공포','패닉','분노','화가','화를','절망','괴롭','예민','짜증','답답','악몽',
       '놀라','놀란','움찔','하얘','떨리','떨린','식은땀','숨이','숨결','호흡','심장','깬다','깨어','잠들',
       '잠을 못','도망','죽고','피하','회피','떠오','떠올','재생','끌려','굳는','멈춘','통제되지',
       '집중할 수 없','집중이','지친','무기력','우울','걱정','부담','긴장','스트레스','힘들','싫','외롭','눈물']
RES = ['지만','했지만','었지만','통해','극복','회복','보람','익숙','괜찮','나아졌','더 이상','않는다','않다',
       '금세','평소대로','차분','침착','자신감','인정', '가능', '좋', '웠던', '소중']

def lexicon_features(texts):
    f = [[sum(w in t for w in NEG), sum(w in t for w in RES), int('지만' in t), len(t) / 100] for t in texts]
    return sparse.csr_matrix(np.asarray(f, dtype=float))

# ---------- 3. 모델 ----------
def model_char_lex():
    feats = make_union(TfidfVectorizer(analyzer="char", ngram_range=(1, 3), sublinear_tf=True),
                       FunctionTransformer(lexicon_features))
    return make_pipeline(feats, LogisticRegression(C=C, class_weight="balanced", max_iter=5000, random_state=SEED))

def model_charwb():
    return make_pipeline(TfidfVectorizer(analyzer="char_wb", ngram_range=(1, 4), sublinear_tf=True),
                         LogisticRegression(C=C, class_weight="balanced", max_iter=5000, random_state=SEED))

MODELS = [model_char_lex, model_charwb]

def fit_predict(X_tr, y_tr, X_te):
    return np.mean([m().fit(X_tr, y_tr).predict_proba(X_te)[:, 1] for m in MODELS], axis=0)

# ---------- 4. 반복 교차검증으로 성능/임계값 확인 ----------
rskf = RepeatedStratifiedKFold(n_splits=5, n_repeats=5, random_state=SEED)
splits = list(rskf.split(X, y)); oofs = []
for r in range(5):
    oof = np.zeros(len(y))
    for tr_idx, va_idx in splits[r*5:(r+1)*5]:
        oof[va_idx] = fit_predict(X[tr_idx], y[tr_idx], X[va_idx])
    oofs.append(oof)

print("threshold | F1(평균 5회 반복)")
for th in [0.36, 0.40, 0.44, 0.48, 0.52]:
    print(f"   {th:.2f}   | {np.mean([f1_score(y, o >= th) for o in oofs]):.4f}")
o = oofs[0] >= THRESHOLD
print(f"[선택 th={THRESHOLD}] P={precision_score(y,o):.3f} R={recall_score(y,o):.3f} F1={f1_score(y,o):.3f}")

# ---------- 5. 전체 학습 → 예측 → 저장 ----------
p_test = fit_predict(X, y, X_test)
sub = pd.DataFrame({"id": test["id"], "stress": (p_test >= THRESHOLD).astype(int)}).sort_values("id")
sub.to_csv(OUT_PATH, index=False)
print(f"saved {OUT_PATH} | 양성 비율 {sub.stress.mean():.3f} (train {y.mean():.3f})")
print(sub.head())
```

```python
import os, sys, gc, warnings, random
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score, precision_recall_fscore_support
from sklearn.pipeline import Pipeline
from sklearn.utils.class_weight import compute_class_weight

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

TRAIN_PATH = "/content/drive/MyDrive/MAICON2025_qualification/problem3/data/train.csv"
TEST_PATH  = "/content/drive/MyDrive/MAICON2025_qualification/problem3/data/test.csv"
OUT_PATH   = "submission.csv"

def read_csv_any(path):
    for enc in ["utf-8-sig", "utf-8", "cp949"]:
        try:
            return pd.read_csv(path, encoding=enc)
        except Exception:
            continue
    return pd.read_csv(path)

train = read_csv_any(TRAIN_PATH)
test  = read_csv_any(TEST_PATH)

assert {"text", "stress"}.issubset(set(train.columns)), "train.csv는 text, stress 컬럼이 필요합니다."
assert {"id", "text"}.issubset(set(test.columns)), "test.csv는 id, text 컬럼이 필요합니다."

train["text"] = train["text"].astype(str).fillna("")
test["text"]  = test["text"].astype(str).fillna("")
y = train["stress"].astype(int).values


vectorizer = TfidfVectorizer(
    analyzer="char",
    ngram_range=(2,5),
    min_df=2,
    max_features=300_000,
    lowercase=False
)

logreg = LogisticRegression(
    C=2.0,
    max_iter=3000,
    class_weight="balanced",
    solver="saga",
    n_jobs=None,
    random_state=SEED
)

pipe = Pipeline([
    ("tfidf", vectorizer),
    ("clf", logreg)
])

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

oof_proba = np.zeros(len(train), dtype=float)
thresholds = []

for fold, (trn_idx, val_idx) in enumerate(skf.split(train["text"], y), 1):
    X_trn = train["text"].iloc[trn_idx].values
    y_trn = y[trn_idx]
    X_val = train["text"].iloc[val_idx].values
    y_val = y[val_idx]

    pipe.fit(X_trn, y_trn)
    if hasattr(pipe.named_steps["clf"], "predict_proba"):
        p_val = pipe.predict_proba(X_val)[:, 1]
    else:
        df = pipe.decision_function(X_val)
        p_val = 1 / (1 + np.exp(-df))

    oof_proba[val_idx] = p_val

    cand = np.linspace(0.1, 0.9, 81)
    best_th, best_f1 = 0.5, -1
    for th in cand:
        pred = (p_val >= th).astype(int)
        f1 = f1_score(y_val, pred, pos_label=1, zero_division=0)
        if f1 > best_f1:
            best_f1 = f1
            best_th = th
    thresholds.append(best_th)
    print(f"[Fold {fold}] best_th={best_th:.2f}, F1(1)={best_f1:.4f}")

global_th = float(np.median(thresholds))
oof_pred = (oof_proba >= global_th).astype(int)
p, r, f1, _ = precision_recall_fscore_support(y, oof_pred, average="binary", pos_label=1, zero_division=0)

print("\n=== OOF (label 1 기준) ===")
print(f"Threshold (median of folds): {global_th:.3f}")
print(f"Precision: {p:.4f} | Recall: {r:.4f} | F1: {f1:.4f}")

pipe.fit(train["text"].values, y)

if hasattr(pipe.named_steps["clf"], "predict_proba"):
    p_test = pipe.predict_proba(test["text"].values)[:, 1]
else:
    df = pipe.decision_function(test["text"].values)
    p_test = 1 / (1 + np.exp(-df))

test_pred = (p_test >= global_th).astype(int)

sub = pd.DataFrame({
    "id": test["id"].values,
    "stress": test_pred.astype(int)
})

sub = sub.sort_values("id").reset_index(drop=True)
sub.to_csv(OUT_PATH, index=False, encoding="utf-8-sig")

print(f"\nSaved: {OUT_PATH}")
print(sub.head())
```

```python
import torch
import pandas as pd
from torch.utils.data import Dataset
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, f1_score
from transformers import (
    RobertaTokenizer,
    RobertaForSequenceClassification,
    Trainer,
    TrainingArguments
)

# ==============================
# 1. 데이터 로딩
# ==============================
train_df = pd.read_csv("/content/drive/MyDrive/MAICON2025_qualification/problem3/data/train.csv")
test_df  = pd.read_csv("/content/drive/MyDrive/MAICON2025_qualification/problem3/data/test.csv")

print("Train shape:", train_df.shape, "Test shape:", test_df.shape)

# ==============================
# 2. 디바이스 설정
# ==============================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
N_EPOCHS = 30

# ==============================
# 3. 토크나이저 & 모델
# ==============================
tokenizer = RobertaTokenizer.from_pretrained("roberta-base")
model_roberta = RobertaForSequenceClassification.from_pretrained(
    "roberta-base",
    num_labels=2
).to(device)

print("RoBERTa model loaded successfully.")

# ==============================
# 4. Dataset 클래스
# ==============================
class TextDataset(Dataset):
    def __init__(self, dataframe, tokenizer, max_len, is_test=False):
        self.data = dataframe.reset_index(drop=True)  # ✅ 안전하게 리셋
        self.tokenizer = tokenizer
        self.max_len = max_len
        self.is_test = is_test

    def __len__(self):
        return len(self.data)

    def __getitem__(self, index):
        text = self.data.iloc[index]["text"]

        encoding = self.tokenizer.encode_plus(
            text,
            add_special_tokens=True,
            max_length=self.max_len,
            return_token_type_ids=False,
            padding="max_length",
            return_attention_mask=True,
            return_tensors="pt",
            truncation=True
        )

        item = {
            "input_ids": encoding["input_ids"].flatten(),
            "attention_mask": encoding["attention_mask"].flatten()
        }
        if not self.is_test:
            label = self.data.iloc[index]["stress"]
            item["labels"] = torch.tensor(label, dtype=torch.long)
        return item


# ==============================
# 5. Train/Validation split
# ==============================
train_df_, val_df = train_test_split(
    train_df,
    test_size=0.2,
    stratify=train_df["stress"],
    random_state=42
)

MAX_LEN = 512
BATCH_SIZE = 16

train_dataset = TextDataset(train_df_, tokenizer, MAX_LEN, is_test=False)
val_dataset   = TextDataset(val_df, tokenizer, MAX_LEN, is_test=False)
test_dataset  = TextDataset(test_df, tokenizer, MAX_LEN, is_test=True)

# ==============================
# 6. TrainingArguments
# ==============================
training_args = TrainingArguments(
    output_dir="./results_roberta",
    num_train_epochs=N_EPOCHS,
    per_device_train_batch_size=BATCH_SIZE,
    per_device_eval_batch_size=BATCH_SIZE,
    warmup_steps=100,
    weight_decay=0.01,
    logging_dir="./logs_roberta",
    logging_steps=10,
    save_steps=200,          # 모델 저장 주기 (step 단위)
    eval_steps=200,          # ✅ 평가 주기 (step 단위, 구버전 호환)
    learning_rate=2e-5,
    report_to=[]             # wandb/tensorboard 로그 방지
)


# ==============================
# 7. Metrics (F1 기준)
# ==============================
def compute_metrics(pred):
    labels = pred.label_ids
    preds = pred.predictions.argmax(-1)
    precision = precision_score(labels, preds, pos_label=1)
    recall = recall_score(labels, preds, pos_label=1)
    f1 = f1_score(labels, preds, pos_label=1)
    return {"precision": precision, "recall": recall, "f1": f1}

# ==============================
# 8. Trainer
# ==============================
trainer = Trainer(
    model=model_roberta,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,    # ✅ eval_dataset 있으면 자동으로 평가 실행
    compute_metrics=compute_metrics,
    tokenizer=tokenizer
)

# ==============================
# 9. 학습
# ==============================
print("Starting RoBERTa model training...")
trainer.train()
print("RoBERTa model training finished.")

# ==============================
# 10. Test 예측
# ==============================
preds = trainer.predict(test_dataset)
labels = preds.predictions.argmax(-1)

# ==============================
# 11. 제출 파일 생성
# ==============================
submission = pd.DataFrame({
    "id": test_df["id"],
    "stress": labels
})
submission = submission.sort_values("id")
submission.to_csv("submission.csv", index=False)

print("✅ submission.csv 생성 완료")
print(submission.head())

# ==============================
# 12. 모델 저장
# ==============================
model_roberta.save_pretrained("./fine_tuned_roberta")
tokenizer.save_pretrained("./fine_tuned_roberta")
print("Fine-tuned RoBERTa model and tokenizer saved.")
```
