# 병사 발언 분석을 통한 스트레스 징후 탐지하기

본 임무의 목적은 주어진 병사의 발언을 분석하여 스트레스 징후를 예측하는 분류기를 만드는 것입니다. 본 과제에서 ‘병사’는 현역과 전역자를 모두 포괄합니다. 스트레스 징후를 파악하는 것이 중요하기에 평가는 단순 전체 정확도가 아닌 스트레스 징후를 나타내는 데이터를 기준으로 F1 점수를 계산하여 진행합니다.

## 훈련 데이터

`train.csv`는 두 개의 필드로 구성되며 총 300개의 레코드가 제공됩니다.

| 필드 | 타입 | 설명 |
| --- | --- | --- |
| text | 문자열 | 병사 발언에 해당하는 짧은 한국어 문장이 들어옵니다. 오타가 존재할 수 있고 띄어쓰기가 안 맞을 수도 있습니다. |
| stress | 숫자 | 예측 대상이 되는 레이블 필드로 1이면 스트레스 징후가 있는 것이고 0이면 정상입니다 |

처음 필드 text는 병사의 말이고 stress는 그 text에 스트레스 가능성을 표현하는지 여부로 1이면 스트레스 가능성이 있다고 간주합니다. 0이면 정상입니다. 아래 처음 5개의 레코드를 예로 보여드렸습니다.

| text | stress |
| --- | --- |
| 금속 부딪히는 소리에 가슴이 철렁해 한동안 걸음을 멈췄다.| 1 |
| 부두 경계 근무 때로는 긴장되지만 밤에 깨더라도 금세 다시 잔다 | 0 |
| 아팠던 몸도 충분한 휴식 후에는 완전히 회복됐다 | 0 |
| 장갑차 운전 교육을 받을 때 큰 차체와 복잡한 조작법 때문에 두려웠지만 교관의 세심한 지도와 충분한 연습을 통해 안전하고 정확한 기동이 가능해졌다 | 0 |
| 그뒤로 훈련 중 폭죽 소리 때로는 긴장되지 만경계는 하되 업무는 평소대로 한다 | 0 |

이 데이터셋을 이용해 직접 분류기 모델을 학습(train) 하고 평가해야 합니다. 단, GPT와 같은 대규모 언어 모델(LLM)을 직접 불러와 Zero-Shot 또는 Few-Shot과 같은 프롬프트 엔지니어링만으로 분류 작업을 수행하는 것은 허용되지 않습니다. 즉, LLM을 단순 분류기로 호출하는 것은 금지하며, **반드시 제공된 데이터셋을 기반으로 별도의 모델을 학습시켜야 합니다.** 모델을 명확히 학습해야 하기에 매뉴얼하게 직접 분류하여 제출하는 것 역시 허용하지 않습니다. 필요시 작성 코드를 확인할 예정입니다.

## 평가 기준

앞서 언급했던 것처럼 스트레스 징후를 찾아내는 것이 더 중요하기에 스트레스를 나타내는 레이블 1(스트레스 징후 있음)을 기준으로 f1 스코어를 계산합니다. 레이블 1을 기준으로 정밀도(precision)와 재현율(recall)을 계산한 다음에 아래와 같은 공식으로 f1 스코어를 계산합니다.

* precision = 모델이 "스트레스 징후 있음" 이라고 예측한 것 중에서 실제로 맞은 비율
* recall = 실제로 "스트레스 징후 있음" 인 것 중에서 모델이 제대로 찾아낸 비율
* f1 = 2 * precision * recall / (precision + recall)

```python
import re
import numpy as np
from sklearn.base import BaseEstimator, TransformerMixin
# ---- 감정/증상 특징 추출기: 학습 데이터(300개 발언)에서 관찰한 스트레스 징후 유형을 사전으로 정리해 수치 특징으로 변환 ----
# 유형: 공포/불안/분노/절망, 신체 반응, 재경험(플래시백/악몽), 과각성/회피, 수면, 피로, 해소/성공 표현, 부정·역접 구조

# 감정/증상 사전 (도메인 지식 기반). 띄어쓰기/오타 대응을 위해 공백 제거 후 부분 문자열 매칭
LEX = {
    'fear':     ['공포', '두렵', '무섭', '무서', '패닉', '놀라', '놀랄', '깜짝', '철렁', '움찔', '겁'],
    'anxiety':  ['불안', '걱정', '초조', '긴장', '예민', '의심', '부담', '압박', '조마조마'],
    'anger':    ['분노', '화가', '화를', '화나', '짜증', '치밀', '억울'],
    'despair':  ['절망', '죽고', '괴롭', '괴로', '우울', '외롭', '도망', '막막', '무기력', '포기', '자책'],
    'body':     ['가슴', '심장', '숨', '호흡', '땀', '손바닥', '몸이', '온몸', '손끝', '흉부', '조여', '떨', '굳', '딱딱', '식은땀', '진동', '두근'],
    'intrusion': ['떠오', '장면', '화면', '재생', '끌려', '겹쳐', '겹쳐보', '머릿속', '과거', '악몽', '나쁜꿈', '그날', '플래시', '생각으로'],
    'vigilance': ['확인', '훑', '자리를비', '피하', '반복', '작은소리', '소리에', '소리만', '소리가', '뒤를', '경계'],
    'sleep':    ['깨', '새벽', '잠', '악몽', '불면'],
    'fatigue':  ['지친', '지쳐', '금방지', '집중', '멍해', '머리가하얘', '하얘', '힘들'],
    # ---- 확장 사전: 일반 스트레스/정신건강 어휘 ----
    'isolation': ['외롭', '외로', '고립', '따돌', '소외', '외톨', '서글', '쓸쓸', '공허', '혼자', '왕따', '아무도', '곁에없', '멀어지'],
    'burnout':   ['피로', '지치', '지쳐', '번아웃', '무기력', '의욕', '에너지', '고갈', '탈진', '소진', '무겁', '기운이', '쉬어도', '나른', '녹초'],
    'insomnia':  ['불면', '잠을설', '잠들기', '잠이안', '잠이오지', '뒤척', '밤잠', '설치', '숙면', '잠못', '잠을이루', '수면'],
    'pressure':  ['압박', '스트레스', '시달', '중압', '질책', '비난', '눈치', '시선', '추궁', '압력', '닦달', '갈굼', '가혹', '괴롭힘'],
    'selfblame': ['자책', '죄책', '자괴', '위축', '자신이없', '자신감이없', '무능', '쓸모', '실패할까', '실수할까', '나때문', '내탓', '부끄', '수치', '열등'],
    'trauma':    ['트라우마', '외상', '상처', '되살아', '기억이', '충격', '후유증', '잊혀지지', '잊을수없'],
    'dysreg':    ['눈물', '울컥', '울음', '감정기복', '폭발', '곤두', '신경질', '신경이', '날카로', '격해', '욱하', '짜증', '울고'],
    'avoidance': ['대인기피', '기피', '회피', '피하게', '피하고', '말을아', '입을닫', '숨고', '숨어', '꺼려', '외면', '자리를피', '만나기싫', '가기싫'],
    'hopeless':  ['무력감', '무력', '허무', '의미가없', '끝내고', '사라지고싶', '살기싫', '견딜수없', '고통', '버거', '벗어날수없', '막다른', '희망이없', '살고싶지'],
    'physical':  ['두통', '소화', '식욕', '입맛', '어지럽', '어지러', '메스꺼', '구토', '체한', '근육통', '몸이무겁', '손떨', '떨림', '숨쉬기', '숨이막', '답답'],
    'worry':     ['걱정', '염려', '조바심', '안절부절', '마음이놓이지', '초조', '불편', '위태', '두려', '무섭', '겁이'],
}
RESOLVE = ['성공', '완수', '보람', '뿌듯', '극복', '해냈', '합격', '완벽', '진정', '괜찮', '사라', '회복', '만족', '자랑',
           '도움', '기여', '성장', '잘해', '금세', '금방잔', '곧', '자신감', '차분', '평소대로', '유지', '버텨', '감동', '효율', '정확',
           '체계', '철저', '제공', '교육', '숙달', '마스터', '향상']
CONTRAST = ['지만', '였지만', '했지만', '더라도', '해도', '에도', '었는데', '그래도']
NEG = ['않', '없', '아니', '사라', '진정', '금세', '곧', '괜찮', '넘긴', '넘기']
START = ['가끔', '요즘', '간혹', '어쩌다', '하필', '그뒤로', '그후로', '요새', '문득']


def norm(t):
    return re.sub(r'\s+', '', str(t))


class EmotionFeatures(BaseEstimator, TransformerMixin):
    """텍스트 -> 감정/증상 카운트, 부정/해소 신호, 길이 등 수치 특징 (학습 불필요, fit 은 no-op)"""
    def fit(self, X, y=None):
        return self

    def _row(self, raw):
        t = norm(raw); f = []
        total = 0; negated_total = 0
        for k, words in LEX.items():
            c = 0; neg = 0
            for w in words:
                for m in re.finditer(re.escape(w), t):
                    c += 1
                    tail = t[m.end(): m.end() + 8]          # 감정어 뒤 8글자 안에 부정/해소 표현이 있으면 부정된 감정
                    if any(n in tail for n in NEG):
                        neg += 1
            f += [c, neg, c - neg]
            total += c; negated_total += neg
        res = sum(t.count(w) for w in RESOLVE)
        con = sum(t.count(w) for w in CONTRAST)
        f += [total, negated_total, total - negated_total, res, con,
              int(any(t.startswith(s) for s in START)), int(any(s in t[:6] for s in START)),
              len(t), len(str(raw).split()), str(raw).count('.') + str(raw).count('…'),
              int(con > 0 and res > 0),                      # '힘들었지만 ... 성공' 형태
              (total - negated_total) / (len(t) + 1) * 100,
              int(total - negated_total >= 2), int(total - negated_total >= 3)]
        return f

    def transform(self, X):
        return np.array([self._row(x) for x in X], dtype=float)
```

```python
import os, random, warnings
warnings.filterwarnings("ignore")
import pandas as pd
from scipy.sparse import hstack, csr_matrix
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import make_pipeline, make_union
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.naive_bayes import ComplementNB
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import f1_score

SEED = 42
random.seed(SEED); np.random.seed(SEED)

# 데이터 경로: Colab(초기코드) 경로 우선, 없으면 로컬 후보 사용
CANDIDATES = [
    "/content/drive/MyDrive/MAICON2025_qualification/problem3/data/",
    "data/", "/content/data/",
]
D = next(p for p in CANDIDATES if os.path.exists(p + "train.csv"))
train = pd.read_csv(D + "train.csv")
test = pd.read_csv(D + "test.csv")
X, y, X_test = train.text.values, train.stress.values, test.text.values
print(train.shape, test.shape, train.stress.value_counts().to_dict())

# ---- 모델 (v10): TF-IDF + 감정특징 결합 선형모델 2종 + 문자 TF-IDF ComplementNB + 감정특징 전용 GBM, 확률 평균 ----
class TfidfPlusEmo(BaseEstimator, ClassifierMixin):
    """TF-IDF(글자 1~4gram + 어절 1~2gram, sparse) 와 표준화한 감정 특징(w 배)을 이어 붙여 선형모델에 입력"""
    def __init__(self, clf, w=3.0):
        self.clf = clf; self.w = w
    def _emo(self, T, fit=False):
        E = EmotionFeatures().transform(T)
        if fit: self.sc_ = StandardScaler().fit(E)
        return np.clip(self.sc_.transform(E), -5, 5) * self.w / np.sqrt(E.shape[1])
    def fit(self, T, y):
        self.tf_ = make_union(TfidfVectorizer(analyzer="char_wb", ngram_range=(1, 4), sublinear_tf=True),
                              TfidfVectorizer(analyzer="word", ngram_range=(1, 2), sublinear_tf=True, token_pattern=r"\S+")).fit(T)
        self.clf_ = self.clf.fit(hstack([self.tf_.transform(T), csr_matrix(self._emo(T, True))]).tocsr(), y)
        self.classes_ = self.clf_.classes_
        return self
    def predict_proba(self, T):
        return self.clf_.predict_proba(hstack([self.tf_.transform(T), csr_matrix(self._emo(T))]).tocsr())

def build():
    return {
        "emo_lr":  TfidfPlusEmo(LogisticRegression(C=10, class_weight="balanced", max_iter=3000), w=3),
        "emo_svm": TfidfPlusEmo(CalibratedClassifierCV(LinearSVC(C=0.5, class_weight="balanced"), cv=3), w=3),
        "cnb":     make_pipeline(TfidfVectorizer(analyzer="char_wb", ngram_range=(1, 4), sublinear_tf=True), ComplementNB(alpha=0.3)),
        "emo_hgb": make_pipeline(EmotionFeatures(), HistGradientBoostingClassifier(
                       max_depth=3, learning_rate=0.05, max_iter=200, class_weight="balanced", random_state=0)),
    }

# ---- 검증: 5-Fold x 3회 반복 층화 CV, 레이블 1 기준 F1 ----
N_REP = 3
rskf = RepeatedStratifiedKFold(n_splits=5, n_repeats=N_REP, random_state=0)
oof = {k: np.zeros((N_REP, len(X))) for k in build()}
for i, (tr, va) in enumerate(rskf.split(X, y)):
    for k, m in build().items():
        oof[k][i // 5, va] = m.fit(X[tr], y[tr]).predict_proba(X[va])[:, 1]

def cv_f1(p, t): return float(np.mean([f1_score(y, (p[r] >= t).astype(int)) for r in range(N_REP)]))
for k in oof: print(f"{k:8s} CV F1@0.5 = {cv_f1(oof[k], .5):.3f}")
ens = np.mean(list(oof.values()), axis=0)
# 임계값: train 양성비율(≈0.173). 확률은 train 사전확률(양성 17%)에 맞춰 보정된 값이므로, 배포 시 양성/음성이 비슷하다고 가정(균형 사전확률)하면
# 결정 임계값이 train 양성비율과 같아진다. 스트레스 징후(양성) 탐지가 더 중요하다는 문제 취지와도 일치하는 재현율 우선 규칙이며, 특정 점수에 맞춰 튜닝한 값이 아님
THR = float(y.mean())
print(f"ensemble(v10) CV F1@0.5 = {cv_f1(ens, 0.5):.3f} | CV F1@THR({THR:.3f}) = {cv_f1(ens, THR):.3f}  (이전 v1 TF-IDF 앙상블: 0.821)")
print("임계값 민감도:", {round(t, 2): round(cv_f1(ens, t), 3) for t in np.arange(0.3, 0.71, 0.1)})

# ---- 전체 train으로 재학습 후 test 예측 ----
proba = np.mean([m.fit(X, y).predict_proba(X_test)[:, 1] for m in build().values()], axis=0)
pred = (proba >= THR).astype(int)
print("test 예측 분포:", np.bincount(pred, minlength=2), "| train 양성비율 %.3f / test 예측 양성비율 %.3f" % (y.mean(), pred.mean()))
```
