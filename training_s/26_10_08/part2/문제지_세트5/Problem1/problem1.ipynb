미국 인구조사 자료로 각 사람의 연소득이 5만 달러를 넘을 **확률**을 예측하세요.

## 데이터
- `data/train.csv` : 34,000명 (income 포함)
- `data/test.csv` : 14,842명 (income 없음)

| 컬럼 | 설명 |
|---|---|
| id | 고유 번호 |
| age | 나이 |
| workclass | 고용 형태 (민간, 자영업, 공무원 등) |
| fnlwgt | 인구조사 표본 가중치 |
| education / education_num | 최종 학력 (문자) / 학력 단계 (숫자) |
| marital_status | 혼인 상태 |
| occupation | 직업군 |
| relationship | 가구 내 관계 (남편, 아내, 자녀 등) |
| race / sex | 인종 / 성별 |
| capital_gain / capital_loss | 자본 이득 / 손실 (USD) |
| hours_per_week | 주당 근무 시간 |
| native_country | 출신 국가 |
| **income** | **1 = 연소득 5만 달러 초과, 0 = 이하 (예측 대상)** |

### 데이터 특징
- 고소득자는 약 24%입니다.
- 결측값이 빈칸이 아니라 **`?` 문자**로 기록되어 있습니다 (workclass, occupation, native_country).
- 같은 정보를 다른 형태로 담은 컬럼이 있습니다.

## 평가
**ROC-AUC** — 높을수록 좋습니다.

- 제출한 값의 **순서**로 계산합니다. 0/1보다 확률을 제출하는 것이 유리합니다.

## 초기 코드 (베이스라인)
아래 베이스라인은 수치형 3개(나이·학력 단계·근무 시간)만으로 로지스틱 회귀 확률을 제출합니다. 동작은 하지만 점수가 낮습니다.

````python
import pandas as pd
from sklearn.linear_model import LogisticRegression
DATA_DIR = "data"
train = pd.read_csv(f"{DATA_DIR}/train.csv"); test = pd.read_csv(f"{DATA_DIR}/test.csv")
num = ["age", "education_num", "hours_per_week"]          # 수치형 3개만 사용
clf = LogisticRegression(max_iter=1000).fit(train[num], train.income)
pd.DataFrame({"id": test.id, "income": clf.predict_proba(test[num])[:, 1]}).to_csv("submission.csv", index=False)
````
