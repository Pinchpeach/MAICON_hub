# Problem 3. 정답 없이 손글씨 숫자 묶기 (군집화)

우체국 자동화 팀이 손글씨 숫자 이미지를 잔뜩 모았지만, **어떤 숫자인지 라벨이 하나도 없습니다.**
라벨 없이 비슷한 이미지끼리 **10개 그룹**으로 묶으세요. 같은 숫자끼리 잘 묶일수록 점수가 높습니다.

## 데이터
- `data/digits.csv` : 1,797장, 학습·테스트 구분 없음

| 컬럼 | 설명 |

| pixel0 ~ pixel63 | 8×8 흑백 이미지를 한 줄로 편 밝기(0~16). `pixel{8*행 + 열}` |

## 평가
**ARI (Adjusted Rand Index)** — 높을수록 좋습니다 (최대 1, 무작위 묶기 ≈ 0).

- 그룹 번호 자체는 상관없습니다. "같은 숫자는 같은 그룹, 다른 숫자는 다른 그룹"이 얼마나 지켜졌는지만 봅니다.
- 그룹 수는 자유지만, 10개가 아니면 점수가 떨어지기 쉽습니다.

## 초기 코드 (베이스라인)
아래 베이스라인은 원본 픽셀 64개로 K-평균 군집화를 합니다. 동작은 하지만 점수가 낮습니다.

````python
import pandas as pd
from sklearn.cluster import KMeans
DATA_DIR = "data"
df = pd.read_csv(f"{DATA_DIR}/digits.csv")
X = df.drop(columns="id").values
labels = KMeans(n_clusters=10, n_init=10, random_state=0).fit_predict(X)   # 원본 픽셀 그대로 K-평균
pd.DataFrame({"id": df.id, "cluster": labels}).to_csv("submission.csv", index=False)
````
