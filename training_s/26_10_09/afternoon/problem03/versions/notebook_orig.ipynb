```python
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, accuracy_score
DATA=Path('data')
train=pd.read_csv(DATA/'train.csv'); test=pd.read_csv(DATA/'test.csv'); sample=pd.read_csv(DATA/'sample_submission.csv')
print(train.shape, test.shape)
```

```python
features=[c for c in test.columns if c!='soldier_id']
X=train[features]; y=train['overall_satisfaction']
a,b=train_test_split(np.arange(len(train)),test_size=.25,random_state=42)
model=RandomForestRegressor(n_estimators=64,n_jobs=2,random_state=42)
model.fit(X.iloc[a],y.iloc[a])
print('Validation RMSE:',mean_squared_error(y.iloc[b],model.predict(X.iloc[b]))**.5)
# TODO: 모형·특징·검증 분할을 비교하세요.
model.fit(X,y)
submission=pd.DataFrame({'soldier_id':test.soldier_id,'overall_satisfaction':model.predict(test[features])})[sample.columns]
submission.to_csv('submission.csv',index=False)
```
