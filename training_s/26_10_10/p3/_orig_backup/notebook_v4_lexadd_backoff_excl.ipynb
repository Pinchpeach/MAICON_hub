## 평가 기준

여섯 가지 고장 유형을 모두 고르게 잘 분류하는 것이 중요하기 때문에 **Macro F1 Score**를 주요 평가 지표로 사용합니다. 각 고장 유형별로 precision과 recall을 이용하여 F1 score를 계산한 뒤, 6개 클래스의 F1 score를 동일한 비중으로 평균하여 Macro F1을 계산합니다.

- **Precision**: 특정 고장 유형이라고 예측한 데이터 중 실제로 해당 고장 유형인 비율
- **Recall**: 실제 특정 고장 유형인 데이터 중 모델이 해당 유형으로 정확히 찾아낸 비율
- **F1 Score**: `2 × Precision × Recall / (Precision + Recall)`
- **Macro F1**: 6개 고장 유형 각각의 F1 Score를 계산한 뒤 단순 평균한 값

# 풀이 노트 (요약)
- **문제:** 정비 보고서(`report`) -> 고장 유형 6클래스, 지표 Macro F1. train 540건(클래스당 90), test 180건.
- **핵심 관찰 (1절):** train 540건은 서로 다른 핵심 증상 문장 93개를 6번씩 다른 도입부·상투구·주변 문장으로 감싼 것이고, test 는 서식이 완전히 다르며 핵심 문장이 train 에 글자 그대로 한 번도 나오지 않는다. 따라서 **핵심 증상 문장 단위로 묶은 그룹 CV**로 평가한다(원문 그대로의 5-Fold 1.000 은 무의미).
- **전처리 (2절):** 도입부·상투구·다른 계통의 "정상" 문장을 걷어내고 핵심 증상 절만 남긴다. 서술 어미는 지운다.
- **모델 (v4, 메인):**
  1. 글자 TF-IDF(char_wb 1~2-gram) + **계통별 용어 일치 수 6개** -> 선형 SVM. 용어 목록은 지문의 계통 정의와 일반 정비 용어로 만든 기본 목록에, 용어가 걸리지 않던 test 핵심 절을 train 핵심 절과 대조해 고른 추가 용어를 더한 것이다(사용자 지시).
  2. 핵심 절에 용어가 하나도 없으면 **글자만 쓰는 모델로 물러난다**(용어 특징에 기댄 모델은 용어가 없을 때 한 클래스로 쏠린다).
  3. "~을 의심했으나 정상", "~을 뒷받침하는 징후는 발견되지 않았다" 같은 문장이 지목한 계통은 **후보에서 뺀다**(train 98건에서 정답을 잘못 뺀 경우 0건).
- **전처리 보강:** 도입부 바로 뒤에 다른 계통 점검 문장이 오는 보고서와 `재현 시험 결과,` 서식을 처리하도록 추출 규칙을 고쳤다(train 핵심 절 93개 -> 90개, 모두 6회 반복).
- **검증의 한계:** 기본 용어는 train 예시를 본 뒤, 추가 용어는 test 문장을 본 뒤 정했다. 따라서 용어를 쓰는 구성의 CV 점수는 낙관적이고, 추가 용어가 test 에서 맞는지는 train 검증으로 확인되지 않는다. 용어가 걸리지 않는 새 표현에 대한 대비로, 검증 쪽 17%에서 용어를 지운 모의 검증을 함께 본다.

## 2. 전처리: 핵심 증상 절 추출

```python
INTRO = [
    r'점검 대상은 [^,]{2,24}이며, 운용자가 ', r'[가-힣0-9 ]{2,24} 정기 점검 중 다음 사항을 확인함\.\s*', r'현장 정비 기록:\s*[가-힣0-9 ]{2,24}?에서 ',
    r'[가-힣0-9 ]{2,24} (?:운용 후 )?정비반 보고:\s*', r'[가-힣0-9 ]{2,24} 기능 시험 결과 ',                                    # train 서식
    r'정비 메모 — [^:]{2,24}:\s*', r'운용자 인계 ?(?:사항|기록)\([^)]{2,24}\):\s*', r'\[현상\]\s*', r'\[점검 메모\]\s*[가-힣0-9 ]{2,24}?에서 ',
    r'[가-힣0-9 ]{2,24}에 대한 재현 시험에서 ', r'[가-힣0-9 ]{2,24} 재현 시험에서 ', r'[가-힣0-9 ]{2,24} 점검 결과,\s*', r'[가-힣0-9 ]{2,24} 재현 시험 결과,\s*', r'[가-힣0-9 ]{2,24} 운용 후 보고:\s*',  # test 서식
]
FILLER = [r'\s*정비 후 첫 운용에서 확인됨', r'\s*증상은 2회 연속 재현됨', r'\s*외기 온도는 정상 범위였음', r'\s*15분 점검 중 동일 현상이 반복됨',
          r'\s*동일 현상이 재시험에서도 반복되었다', r'\s*짧은 확인 운전에서도 같은 현상이 나타났다', r'\s*주변 환경 조건은 정상 범위였다',
          r'\s*정지 후 다시 시작하면 일시적으로 정상처럼 보인다', r'(?:이라고|라고) 보고함', r'\s*이 확인됨']
DISTRACTOR = r'반복 점검했지만|함께 확인했으나|별도 점검에서|의심했으나|우선 점검한 결과|뒷받침하는 징후|정비반은|유사했지만|이상 가능성은 낮|별도 확인 항목|동시에 점검|한편|참고로|추가 확인 결과'
_intro = [re.compile(p) for p in INTRO]; _fill = [re.compile(p) for p in FILLER]
_end = re.compile(r'(했다|였다|었다|았다|되었다|됐다|이다|한다|된다|않는다|않았다|없다|있다|다|함|임|음|됨|했음|였음|됐음)$')

def core_clause(text):
    t = re.sub(r'\s+', ' ', text.strip()); best = None
    for rx in _intro:
        for m in rx.finditer(t):
            if best is None or m.start() < best.start() or (m.start() == best.start() and m.end() > best.end()): best = m
    if best is not None:
        segs = [x for x in re.split(r'(?<=[가-힣다])\.\s|\.$|\s\[대상\]|\s\[기타\]', t[best.end():]) if x and x.strip()]
        seg = next((x for x in segs if not re.search(DISTRACTOR, x)), segs[0] if segs else t)      # 도입부 바로 뒤가 다른 계통 점검 문장이면 건너뛴다
    else:                                                    # 도입부를 못 찾으면 주변 문장만 지운 나머지
        keep = [s for s in re.split(r'(?<=\.)\s+', t) if s and not re.search(DISTRACTOR, s)]
        seg = ' '.join(keep) if keep else t
    for rx in _fill: seg = rx.sub('', seg)
    return seg.strip(' .')

for d in (train, test):
    d['core'] = d.report.map(core_clause); d['core_n'] = d.core.map(lambda s: _end.sub('', s.strip()))
print('핵심 절 길이 평균: train %.0f / test %.0f 글자 | 빈 절: %d, %d' % (train.core.str.len().mean(), test.core.str.len().mean(), (train.core.str.len() < 8).sum(), (test.core.str.len() < 8).sum()))
vc = train.core.value_counts()
print('train 의 서로 다른 핵심 절 %d개 / %d건 | 반복 횟수 분포 %s' % (train.core.nunique(), len(train), dict(vc.value_counts().sort_index())))
print('같은 핵심 절인데 라벨이 다른 경우:', int((train.groupby('core').fault_type.nunique() > 1).sum()), '| test 핵심 절 %d개 / %d건 | train 과 글자 그대로 겹치는 test 핵심 절: %d건' % (test.core.nunique(), len(test), test.core.isin(set(train.core)).sum()))
for lab, r in train.groupby('fault_type').head(1).iterrows(): print('  ', r.fault_type, '|', r.core)
print('test 예:', test.core.iloc[:3].tolist())

# ---- 계통별 용어(지문의 계통 정의 + 일반 정비 용어. 고정 목록이며 test 문장을 보고 늘리지 않는다) ----
CLASSES = ['COMMUNICATION', 'COOLING', 'ELECTRICAL', 'ENGINE', 'HYDRAULIC', 'MOBILITY']
LEX = {'COMMUNICATION': '통신|송신|수신|교신|무전|무선|신호|안테나|주파수|음성|잡음|링크|채널',
       'COOLING': '냉각|라디에이터|팬|온도|과열|방열|수온|서모스탯',
       'ELECTRICAL': '전기|전원|전압|전류|배터리|배선|퓨즈|회로|단자|커넥터|조명|램프|계기|충전|발전기|스위치|표시',
       'ENGINE': '엔진|동력|시동|기동|출력|회전|연료|배기|공회전|흡기|가속|점화|매연',
       'HYDRAULIC': '유압|압력|작동유|실린더|펌프|호스|피팅|누유|밸브|유량|붐|리프트|유체',
       'MOBILITY': '주행|조향|제동|브레이크|타이어|바퀴|핸들|차체|현가|궤도|변속|클러치|차축|휠'}
# 추가 용어(2026-10-10, 사용자 지시): 용어가 걸리지 않던 test 핵심 절과 train 핵심 절을 대조해 더했다. test 문장을 보고 정한 것이므로 이 용어의 효과는 train 검증으로 확인되지 않는다.
LEX_ADD = {'COMMUNICATION': '상대|단말|발신|호출|접속|메시지|데이터|전송|패킷|세션|자료 교환|시간 초과|네트워크|통화',
           'COOLING': '뜨거|식힌|식히|열 축적|보호 경보|보호 모드|보호 기능|흡입부|흡입구|배출구|풍량|연속 운전|장시간 운전|계속 상승',
           'ELECTRICAL': '축전|전력|차단기|표시등|백라이트',
           'ENGINE': '거친|힘이 빠|윤활|노킹|오일|재시동',
           'HYDRAULIC': '습윤|젖어|액추에이터|저압|구동부|구동 장치|구동이|구동 명령|계통 압력|압력계|배관|액면',
           'MOBILITY': '차륜|직진|한쪽으로|쏠림|페달|구동축|공기압|일정 속도|속도가 높|감속'}
USE_ADD = True
if USE_ADD: LEX = {c: LEX[c] + '|' + LEX_ADD[c] for c in LEX}
_lex = {c: re.compile(p) for c, p in LEX.items()}
def lex_counts(texts): return np.array([[len(_lex[c].findall(t)) for c in CLASSES] for t in texts], float)

# ---- 주변 문장이 "이 계통은 아니다"라고 밝힌 경우(강한 문형만 사용) ----
SYS = {'ENGINE': '엔진|동력 계통|연료|회전수|출력', 'ELECTRICAL': '전원|배터리|배선|전압|조명|퓨즈|커넥터|단자', 'HYDRAULIC': '유압|압력|누유|유량',
       'COMMUNICATION': '통신|송수신|송신|수신|무전기|무선|연결 장애|호출', 'MOBILITY': '주행|조향|제동|타이어', 'COOLING': '냉각|팬 성능|온도와|운전 온도'}
STRONG = re.compile('의심했으나|우선 점검한 결과|뒷받침하는 징후|유사했지만|반복 점검했지만|이상 가능성은 낮|함께 확인했으나')
def excluded(report, core):
    out = set()
    for s in re.split(r'(?<=[.])\s+', report.replace(core, ' ')):
        if STRONG.search(s): out |= {k for k, p in SYS.items() if re.search(p, s)}
    return out
for d in (train, test):
    d['ex'] = [excluded(r, c) for r, c in zip(d.report, d.core)]; d['lex_hit'] = lex_counts(d.core_n).sum(1) > 0
viol = sum(l in e for l, e in zip(train.fault_type, train.ex))
print('용어가 하나도 없는 핵심 절: train %d/%d개, test %d/%d개' % ((~train.drop_duplicates('core_n').lex_hit).sum(), train.core_n.nunique(), (~test.drop_duplicates('core_n').lex_hit).sum(), test.core_n.nunique()))
print('배제 정보가 있는 보고서: train %d건(정답을 뺀 경우 %d건), test %d건' % ((train.ex.map(len) > 0).sum(), viol, (test.ex.map(len) > 0).sum()))
print('도입부가 남은 핵심 절(":" 포함): train %d, test %d' % (train.core.str.contains(':').sum(), test.core.str.contains(':').sum()))
```

## 3. 검증: 핵심 절 그룹 5-Fold (새 표현에 대한 일반화를 흉내 낸다)
용어가 걸리지 않는 표현이 test 에 있으므로, 검증 쪽 핵심 절의 17%에서 용어를 지운 모의 검증을 함께 본다.

```python
from scipy.sparse import hstack, csr_matrix
y = train.fault_type.values; yi = np.array([CLASSES.index(l) for l in y]); g = pd.factorize(train.core)[0]

class CoreModel:
    """글자 TF-IDF 모델(char) + 글자·용어 모델(lexm). 용어가 없는 절은 char 로 물러나고, 배제된 계통은 후보에서 뺀다."""
    def fit(self, texts, yy):
        texts = np.asarray(texts, dtype=object); self.v = TfidfVectorizer(analyzer='char_wb', ngram_range=(1, 2), sublinear_tf=True, min_df=1).fit(texts); A = self.v.transform(texts)
        self.char = LinearSVC(C=0.3).fit(A, yy); self.lexm = LinearSVC(C=0.3).fit(hstack([A, csr_matrix(lex_counts(texts))]).tocsr(), yy); return self
    def scores(self, texts, mode='backoff', L=None):
        texts = np.asarray(texts, dtype=object); B = self.v.transform(texts); L = lex_counts(texts) if L is None else L
        s0 = self.char.decision_function(B)
        if mode == 'char': return s0
        s1 = self.lexm.decision_function(hstack([B, csr_matrix(L)]).tocsr())
        return s1 if mode == 'lex' else np.where((L.sum(1) > 0)[:, None], s1, s0)

def apply_exclusion(S, ex):
    S = S.copy()
    for i, e in enumerate(ex):
        for k in e: S[i, CLASSES.index(k)] = -9.0
    return S

NAMES = ['v1 글자만', 'v2a 글자+용어(물러나기 없음)', 'v2 글자+용어, 용어 없으면 글자 모델', 'v4 글자+용어(추가 포함), 물러나기, 계통 배제  <- 메인']
def one_run(sd, drop=0.0):
    rng = np.random.default_rng(sd); gg = pd.Series(g).map(dict(zip(np.unique(g), rng.permutation(len(np.unique(g)))))).values
    miss = rng.random(len(np.unique(g))) < drop                                  # 용어를 지울 핵심 절(모의 검증)
    P = {n: np.zeros(len(train), int) for n in NAMES}
    for a, b in GroupKFold(5).split(train, yi, gg):
        m = CoreModel().fit(train.core_n.values[a], yi[a]); xb = train.core_n.values[b]; L = lex_counts(xb); L[miss[g[b]]] = 0; exb = train.ex.values[b]
        P[NAMES[0]][b] = m.scores(xb, 'char').argmax(1); P[NAMES[1]][b] = m.scores(xb, 'lex', L).argmax(1)
        S = m.scores(xb, 'backoff', L); P[NAMES[2]][b] = S.argmax(1); P[NAMES[3]][b] = apply_exclusion(S, exb).argmax(1)
    return {n: f1_score(yi, p, average='macro') for n, p in P.items()}
t1 = time.time(); A = pd.DataFrame([one_run(sd) for sd in range(5)]); B = pd.DataFrame([one_run(sd, 0.17) for sd in range(5)])
cv_table = pd.DataFrame({'Macro F1': A.mean(), 'sd(5회)': A.std(), '용어 17% 누락 모의': B.mean(), 'sd(5회) ': B.std()}); print(cv_table.round(3).to_string())
print('핵심 절 그룹 5-Fold, 그룹 순서를 바꿔 5회 반복 | %.0f초' % (time.time() - t1))
print('주의: 용어 목록은 train·test 문장을 본 뒤 정했으므로 용어를 쓰는 구성의 점수는 낙관적이다. 오른쪽 열은 용어가 걸리지 않는 새 표현이 17% 있다고 가정한 모의 검증이다.')
```

## 4. 전체 train 으로 재학습, test 예측과 제출 저장

```python
model = CoreModel().fit(train.core_n.values, yi); xt = test.core_n.values
p1 = model.scores(xt, 'char').argmax(1); S = model.scores(xt, 'backoff'); p2 = S.argmax(1); p3 = apply_exclusion(S, test.ex.values).argmax(1)
lab = lambda p: np.array(CLASSES)[p]
Path('versions').mkdir(exist_ok=True)
for name, p in (('v4a_core_charonly_extractfix', p1), ('v4b_core_lexadd_backoff', p2), ('v4_core_lexadd_backoff_excl', p3)):
    pd.DataFrame({'id': test.id.values, 'fault_type': lab(p)}).to_csv('versions/%s.csv' % name, index=False)
sub = pd.DataFrame({'id': test.id.values, 'fault_type': lab(p3)})                 # 메인 = v4
for name, p in (('v1 글자만', p1), ('v4b +용어·물러나기', p2), ('v4 +배제(메인)', p3)): print('%-18s test 예측 분포 %s' % (name, dict(zip([c[:4] for c in CLASSES], np.bincount(p, minlength=6).tolist()))))
print('test 가 클래스당 30건이면 완전 균형 | 예측 일치율 글자만-v4 %.3f, v4b-v4 %.3f' % ((p1 == p3).mean(), (p2 == p3).mean()))
print('글자 모델로 물러난 test 보고서 %d건 | 배제 규칙이 바꾼 test 예측 %d건' % ((~test.lex_hit).sum(), (p2 != p3).sum()))
split = pd.Series(lab(p3)).groupby(test.core_n.values).nunique(); print('같은 핵심 절인데 예측이 갈리는 묶음 %d개(배제 정보가 보고서마다 달라서 생김)' % (split > 1).sum())
```
