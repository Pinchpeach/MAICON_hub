# Problem1 — 군용 항공기와 일반 항공기 탐지·계수

한 이미지에서 모든 객체를 찾아 바운딩박스와 클래스를 제출하고, 클래스별 수를 세십시오. 클래스는 `0`과 `1`입니다.

## 파일 구조
`data/train/images`, `train_images.csv`, `train_annotations.csv`, `test/images`, `test_images.csv`, `sample_submission.csv`

## 제출
`submission.csv` 컬럼: `image_id,class_id,confidence,xmin,ymin,xmax,ymax`. 좌표는 픽셀 기준이고 confidence는 0~1입니다.

## 평가
주 지표는 mAP@0.50, 보조 지표는 클래스별 Count MAE입니다. Private 60%로 최종 순위를 정합니다. 원본 파일명·URL·정답 조회는 금지합니다.

## 데이터 규모
train 이미지 156, train 상자 246, test 이미지 38. Public 16, Private 22.
